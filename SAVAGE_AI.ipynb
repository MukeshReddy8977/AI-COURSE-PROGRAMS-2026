### 3. Install and Run Ollama in Colab
Ollama runs as a background service. We will install it, start the server in the background, and pull the required `llama3.2` model.

```python
# Install zstd required by the Ollama install script, then install Ollama
!sudo apt-get update && sudo apt-get install -y zstd
!curl -fsSL https://ollama.com/install.sh | sh

# Start Ollama server in the background and log to a file
import subprocess
import time

print("Starting Ollama server...")
process = subprocess.Popen(["ollama", "serve"], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
time.sleep(5)  # Wait for server to initialize

# Pull the specific model used in ollama_client.py
print("Pulling llama3.2 model (this may take a couple of minutes)...")
!ollama pull llama3.2:latest
```
