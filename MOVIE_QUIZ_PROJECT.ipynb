```python
import requests
import random

API_KEY = "abcaca63"
url = "https://www.omdbapi.com/"

movies = [
    "Inception",
    "Interstellar",
    "The Dark Knight",
    "Titanic",
    "Avatar",
    "Gladiator",
    "Joker",
    "The Matrix"
]

movie = random.choice(movies)

parameters = {
    "apikey": API_KEY,
    "t": movie
}

response = requests.get(url, params=parameters)
data = response.json()

print("🎬 MOVIE QUIZ")
print("Year:", data["Year"])
print("Genre:", data["Genre"])
print("Plot:", data["Plot"])

answer = input("\nGuess the movie: ")

if answer.lower() == data["Title"].lower():
    print("✅ Correct!")
else:
    print("❌ Wrong!")
    print("The movie was:", data["Title"])

print("IMDb Rating:", data["imdbRating"])
```
