```python
import requests
import random

API_KEY = "abcaca63"

mood = input("Enter your mood (happy, action, horror, romantic): ")

movies = {
    "happy": ["The Mask", "Home Alone", "Mr. Bean's Holiday"],
    "action": ["John Wick", "Gladiator", "The Dark Knight"],
    "horror": ["The Conjuring", "Insidious", "It"],
    "romantic": ["Titanic", "The Notebook", "La La Land"]
}

movie = random.choice(movies.get(mood.lower(), ["Inception"]))

url = "https://www.omdbapi.com/"

parameters = {
    "apikey": API_KEY,
    "t": movie
}

response = requests.get(url, params=parameters)

print(response.status_code)
print(response.json())
```
