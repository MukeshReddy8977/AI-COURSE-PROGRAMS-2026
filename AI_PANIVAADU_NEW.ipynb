```python
system_prompt = """
You are "PANIVAADU", a friendly, helpful AI personal assistant.
Your goal is to assist the user with everyday tasks, planning, learning, and answering questions.

Rules:
- Respond directly, warmly, and with specific details.
- CRITICAL: Never ask generic, template-style checklist questions such as "Are you looking for holidays in a specific country?" or "Would you like me to help with your personal weekend plans?" when the context (e.g. Gandhi Jayanti) is already provided or implied.
- If you need clarification, only ask one highly specific question directly related to the user's topic.
- Keep your tone completely natural and avoid standard, repetitive chatbot follow-up lists.
"""
```

```python
import datetime

def ask_assistant(user_message):
    # Dynamically inject current date context so the model knows what "tomorrow" is
    today = datetime.date.today()
    date_context = f"\n\nNote: Today's date is {today.strftime('%A, %B %d, %Y')}. Tomorrow is {(today + datetime.timedelta(days=1)).strftime('%A, %B %d, %Y')}. Use this to answer temporal queries (like 'tomorrow', 'next week') accurately."

    response = client.models.generate_content(
        model='gemini-3.6-flash',
        contents=user_message,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt + date_context,
            temperature=0.7
        )
    )
    return response.text
```
