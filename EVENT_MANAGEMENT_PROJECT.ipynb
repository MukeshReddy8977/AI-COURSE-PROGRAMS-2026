```python
import json
import random

events=[]

# INPUT INSTRUCTIONS:
# Date format: DD/MM/YYYY
# Example: 09/09/2026
# Time format: HH:MM
# Example: 05:00
# Phone number: Enter 10 digits
# Example: 9876543210
# Maximum people: Enter numbers only
# Example: 500
# For Session, Food and Photographer:
# Enter the number shown in the options.

def save_data():
    with open("events.json","w") as f:
        json.dump(events,f)

def load_data():
    global events
    try:
        with open("events.json","r") as f:
            events=json.load(f)
    except FileNotFoundError:
        events=[]

def generate_event_id():
    while True:
        event_id="#"+str(random.randint(100,999))
        found=False
        for event in events:
            if event["id"]==event_id:
                found=True
                break
        if not found:
            return event_id

def validate_date(date):
    if len(date)==10 and date[2]=="/" and date[5]=="/":
        return True
    else:
        return False

def validate_time(time):
    if len(time)==5 and time[2]==":":
        return True
    else:
        return False

def add_event():
    name=input("Enter event name:")
    date=input("Enter event date:")
    time=input("Enter event time:")
    print("1. Morning")
    print("2. Afternoon")
    print("3. Night")
    session_choice=input("Enter session choice:")
    if session_choice=="1":
        session="Morning"
    elif session_choice=="2":
        session="Afternoon"
    elif session_choice=="3":
        session="Night"
    else:
        print("Invalid session choice.")
        return
    venue=input("Enter event venue:")
    organizer=input("Enter organizer name:")
    phone=input("Enter phone number:")
    print("1. Yes")
    print("2. No")
    food_choice=input("Food required? Enter choice:")
    if food_choice=="1":
        food="Yes"
    elif food_choice=="2":
        food="No"
    else:
        print("Invalid food choice.")
        return
    maximum_people=input("Enter maximum number of people:")
    if not validate_date(date):
        print("Invalid date.")
        return
    if not validate_time(time):
        print("Invalid time.")
        return
    if len(phone)!=10 or not phone.isdigit():
        print("Invalid phone number.")
        return
    if not maximum_people.isdigit():
        print("Invalid number of people.")
        return
    print("1. Yes")
    print("2. No")
    photographer_choice=input("Photographer needed? Enter choice:")
    if photographer_choice=="1":
        photographer="Yes"
    elif photographer_choice=="2":
        photographer="No"
    else:
        print("Invalid photographer choice.")
        return
    event_id=generate_event_id()
    event={
        "name":name,
        "id":event_id,
        "date":date,
        "time":time,
        "session":session,
        "venue":venue,
        "organizer":organizer,
        "phone":phone,
        "food":food,
        "maximum_people":maximum_people,
        "photographer":photographer
    }
    events.append(event)
    save_data()
    print("Event added successfully!")
    print("Your Event ID is:",event_id)

def view_events():
    if len(events)==0:
        print("No events found.")
    else:
        for event in events:
            print("================")
            print("Event ID:",event["id"])
            print("Event Name:",event["name"])
            print("Date:",event["date"])
            print("Time:",event["time"])
            print("Session:",event["session"])
            print("Venue:",event["venue"])
            print("Organizer:",event["organizer"])
            print("Phone:",event["phone"])
            if event["food"]=="Yes":
                if event["session"]=="Morning":
                    print("Food: Breakfast")
                elif event["session"]=="Afternoon":
                    print("Food: Lunch")
                else:
                    print("Food: Dinner")
            else:
                print("Food: No food")
            print("Maximum People:",event["maximum_people"])
            print("Photographer Needed:",event["photographer"])

def search_event():
    event_id=input("Enter event ID to search:")
    for event in events:
        if event["id"].lower()==event_id.lower():
            print("================")
            print("Event ID:",event["id"])
            print("Event Name:",event["name"])
            print("Date:",event["date"])
            print("Time:",event["time"])
            print("Session:",event["session"])
            print("Venue:",event["venue"])
            print("Organizer:",event["organizer"])
            print("Phone:",event["phone"])
            if event["food"]=="Yes":
                if event["session"]=="Morning":
                    print("Food: Breakfast")
                elif event["session"]=="Afternoon":
                    print("Food: Lunch")
                else:
                    print("Food: Dinner")
            else:
                print("Food: No food")
            print("Maximum People:",event["maximum_people"])
            print("Photographer Needed:",event["photographer"])
            return
    print("Event not found.")

def update_event():
    event_id=input("Enter event ID to update:")
    for event in events:
        if event["id"].lower()==event_id.lower():
            name=input("Enter new event name:")
            date=input("Enter new date:")
            time=input("Enter new time:")
            print("1. Morning")
            print("2. Afternoon")
            print("3. Night")
            session_choice=input("Enter session choice:")
            if session_choice=="1":
                session="Morning"
            elif session_choice=="2":
                session="Afternoon"
            elif session_choice=="3":
                session="Night"
            else:
                print("Invalid session choice.")
                return
            venue=input("Enter new venue:")
            organizer=input("Enter new organizer name:")
            phone=input("Enter new phone number:")
            print("1. Yes")
            print("2. No")
            food_choice=input("Food required? Enter choice:")
            if food_choice=="1":
                food="Yes"
            elif food_choice=="2":
                food="No"
            else:
                print("Invalid food choice.")
                return
            maximum_people=input("Enter new maximum number of people:")
            print("1. Yes")
            print("2. No")
            photographer_choice=input("Photographer needed? Enter choice:")
            if photographer_choice=="1":
                photographer="Yes"
            elif photographer_choice=="2":
                photographer="No"
            else:
                print("Invalid photographer choice.")
                return
            if not validate_date(date):
                print("Invalid date.")
                return
            if not validate_time(time):
                print("Invalid time.")
                return
            if len(phone)!=10 or not phone.isdigit():
                print("Invalid phone number.")
                return
            if not maximum_people.isdigit():
                print("Invalid number of people.")
                return
            event["name"]=name
            event["date"]=date
            event["time"]=time
            event["session"]=session
            event["venue"]=venue
            event["organizer"]=organizer
            event["phone"]=phone
            event["food"]=food
            event["maximum_people"]=maximum_people
            event["photographer"]=photographer
            save_data()
            print("Event updated successfully!")
            print("Event ID:",event["id"])
            return
    print("Event not found.")

def delete_event():
    event_id=input("Enter event ID to delete:")
    for event in events:
        if event["id"].lower()==event_id.lower():
            events.remove(event)
            save_data()
            print("Event deleted successfully!")
            return
    print("Event not found.")

def main():
    load_data()
    print("========================================")
    print("     WELCOME TO MUKESH EVENT MANAGEMENT")
    print("========================================")
    while True:
        print("\nEvent Management Menu:")
        print("1. Add Event")
        print("2. View Events")
        print("3. Search Event")
        print("4. Update Event")
        print("5. Delete Event")
        print("6. Exit")
        choice=input("Enter your choice (1-6):")
        if choice=="1":
            add_event()
        elif choice=="2":
            view_events()
        elif choice=="3":
            search_event()
        elif choice=="4":
            update_event()
        elif choice=="5":
            delete_event()
        elif choice=="6":
            print("Thank you for using Mukesh Event Management.")
            break
        else:
            print("Invalid choice. Please try again.")

main()
```
